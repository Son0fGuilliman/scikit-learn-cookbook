```python
import warnings
warnings.filterwarnings("ignore")
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display
from sklearn.metrics import classification_report, accuracy_score

def show_report(y_true, y_pred, target_names=None, **kw):
    """Classification report -> DataFrame berwarna (menggantikan blok styling yang diulang di buku)."""
    rep = pd.DataFrame(classification_report(y_true, y_pred, target_names=target_names,
                                             output_dict=True, **kw)).transpose()
    return (rep.style
            .background_gradient(cmap="Blues", subset=["precision", "recall", "f1-score"])
            .format({"precision": "{:.3f}", "recall": "{:.3f}", "f1-score": "{:.3f}", "support": "{:.0f}"}))

import sklearn
print('scikit-learn:', sklearn.__version__)
```

## 1. Introduction to Decision Trees

### Konsep
- **Decision tree** membagi data secara **rekursif** berdasarkan nilai fitur: tiap **node internal** berisi satu keputusan (fitur ≤ ambang), tiap **leaf** berisi prediksi. Dapat dipakai untuk klasifikasi dan regresi, mudah diinterpretasi, dan tidak membutuhkan scaling.
- **Kriteria pemisahan:** pilih split yang membuat subkelompok paling **homogen**.
  - **Gini impurity** (default scikit-learn): $G = 1 - \sum_{k} p_k^2$, dengan $p_k$ proporsi kelas $k$ di node. $G=0$ berarti node murni.
  - **Entropy**: $H = -\sum_k p_k \log_2 p_k$. Pemilihan split memaksimalkan penurunan impurity (*information gain*).
- **Membaca node** pada `plot_tree`: *fitur ≤ ambang* (aturan split), `gini` (kemurnian), `samples` (proporsi data latih di node), `value` (proporsi tiap kelas), `class` (kelas mayoritas, yaitu prediksi bila node menjadi leaf).
- **Overfitting:** tanpa batasan, pohon tumbuh sampai semua leaf murni sehingga menghafal data latih. Dikendalikan dengan hyperparameter (tabel di bawah).

> **Catatan koreksi:** buku menyebut Gini bernilai antara 0,0 dan 1,0 dengan 1,0 "keseimbangan semua kelas". Nilai maksimum sebenarnya $1 - 1/K$ untuk $K$ kelas: **0,5 untuk 2 kelas dan ≈0,667 untuk 3 kelas** (node akar Iris di bawah ber-gini 0,665, hampir maksimum).

| Hyperparameter | Fungsi |
|---|---|
| `max_depth` | Batas kedalaman pohon; trade-off kompleksitas vs kemurnian node |
| `min_samples_split` | Node dengan sampel lebih sedikit dari nilai ini tidak dibelah lagi |
| `min_samples_leaf` | Minimum sampel di tiap leaf; split yang menyisakan leaf lebih kecil dibatalkan |
| `max_features` | Jumlah fitur yang dipertimbangkan di tiap split (default `None` = semua) |
| `min_impurity_decrease` | Split hanya dilakukan bila penurunan impurity cukup besar |

### Kode

```python
# [Buku] Getting ready
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier

iris = load_iris()
X = iris.data
y = iris.target
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=2024)
print("Train:", X_train.shape, "| Test:", X_test.shape)
```

```python
# [Buku] How to do it: latih dan evaluasi decision tree
clf = DecisionTreeClassifier(random_state=2024)
clf.fit(X_train, y_train)
y_pred = clf.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy:.3f}")        # buku: 0.867
print("Kedalaman pohon:", clf.get_depth(), "| Jumlah leaf:", clf.get_n_leaves())
show_report(y_test, y_pred)
```

```python
# [Buku] How it works: visualisasi pohon
from sklearn.tree import plot_tree

plt.figure(figsize=(15, 10))
plot_tree(clf, feature_names=iris.feature_names, class_names=iris.target_names, filled=True)
plt.title("Decision Tree for Iris Dataset")
plt.show()
```

### Hasil dan interpretasi
- Akurasi **0,867** (sama dengan Figure 8.1). Pohon berkedalaman 4 dengan 5 leaf. Kelas 0 sempurna; kelas 1 dan 2 masih bercampur (precision 0,714 dan 0,846).
- **Membaca pohon** (Figure 8.2): akar membelah pada `petal width ≤ 0,8`; cabang *True* langsung menjadi leaf murni berisi *setosa* (gini 0, sekitar 30,5% data latih). Cabang *False* dibelah lagi oleh `petal length ≤ 4,85` (leaf murni *versicolor*), lalu `petal width ≤ 1,75`, lalu `sepal length ≤ 6,5`. Gini akar 0,665 (distribusi kelas [0,305, 0,362, 0,333]) hampir maksimum untuk 3 kelas.

## 2. Random Forests and Bagging

### Konsep
- **Bagging** (*bootstrap aggregating*): melatih banyak model pada **sampel bootstrap** (diambil **dengan pengembalian**) dari data latih, lalu menggabungkan prediksinya (voting mayoritas untuk klasifikasi, rata-rata untuk regresi). Hasilnya **menurunkan variansi**.
- **Random forest** = bagging pohon keputusan **+ pemilihan fitur acak** pada tiap split (`max_features`; default `RandomForestClassifier` = `"sqrt"`), sehingga pohon-pohonnya makin beragam (tidak saling berkorelasi) dan ensemble makin stabil.
- Satu sampel bootstrap memuat sekitar 63% sampel unik; sisanya (*out-of-bag*) dapat dipakai untuk validasi (`oob_score=True`).
- **Feature importance** (`feature_importances_`): seberapa besar tiap fitur menurunkan impurity, dirata-rata di seluruh pohon. Hati-hati: dihitung pada data latih dan cenderung bias ke fitur bernilai banyak. Random forest kadang dipakai hanya untuk **seleksi fitur** sebelum model lain.
- Bagging dapat diterapkan ke model lain lewat `BaggingClassifier()` dan `BaggingRegressor()`. Hyperparameter utama: `n_estimators`, `max_depth`, `max_features`.

### Kode

```python
# [Buku] Random forest
from sklearn.ensemble import RandomForestClassifier

rf_clf = RandomForestClassifier(n_estimators=100, random_state=2024)
rf_clf.fit(X_train, y_train)
y_pred = rf_clf.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy:.3f}")        # buku: 0.911 (pohon tunggal: 0.867)
show_report(y_test, y_pred)
```

```python
# [Buku] How it works: feature importance
importances = rf_clf.feature_importances_
indices = np.argsort(importances)[::-1]

plt.figure(figsize=(10, 6))
plt.title("Feature Importances in Random Forest")
plt.bar(range(X.shape[1]), importances[indices], align="center")
plt.xticks(range(X.shape[1]), [iris.feature_names[i] for i in indices], rotation=45)
plt.ylabel("Importance")
plt.show()
print(dict(zip([iris.feature_names[i] for i in indices], importances[indices].round(3))))
```

### Hasil dan interpretasi
- Akurasi **0,911** vs pohon tunggal 0,867 (Figure 8.3): selisih **2 sampel uji** dari 45 (41 vs 39 benar). Arahnya sesuai teori (variansi turun), tetapi selisih sekecil itu belum cukup sebagai bukti.
- Feature importance: `petal length` (0,449) dan `petal width` (0,436) menyumbang ≈ 89% total, sedangkan `sepal length` 0,090 dan `sepal width` 0,025 (Figure 8.4). Ini konsisten dengan pairplot Iris di Chapter 6 dan dengan pohon tunggal yang membelah terutama pada fitur petal.

## 3. Gradient Boosting Machines

### Konsep
- **Boosting:** membangun model lemah (umumnya pohon dangkal) **secara berurutan**, dan setiap model baru berusaha memperbaiki kesalahan model sebelumnya. Berbeda dengan random forest yang membangun pohon **paralel** dan independen.
- **Gradient boosting:** model akhir adalah penjumlahan bertahap $F_m(x) = F_{m-1}(x) + \nu\, h_m(x)$, dengan $h_m$ dilatih untuk memprediksi **residual** (lebih tepatnya gradien negatif loss) dari $F_{m-1}$, dan $\nu$ adalah **`learning_rate`**. Mulai dari prediksi awal sederhana lalu ditambah pohon demi pohon.
- Boosting terutama **menurunkan bias**; learning rate kecil butuh lebih banyak pohon (`n_estimators`) tetapi cenderung generalisasi lebih baik. Hyperparameter penting: `learning_rate`, `max_depth`, `n_estimators`.
- **Early stopping:** menghentikan penambahan pohon bila skor validasi berhenti membaik (di scikit-learn: `n_iter_no_change` dan `validation_fraction`). *Catatan:* contoh buku tentang "membatasi pohon lewat hyperparameter" sebenarnya adalah **pre-pruning**; early stopping yang sesungguhnya menghentikan proses *boosting*.
- **Validation curve:** plot skor latih vs validasi terhadap satu hyperparameter. Skor rendah di keduanya → *underfitting*; skor latih tinggi tetapi validasi rendah → *overfitting*.

### Kode

```python
# [Buku] Gradient boosting (1000 pohon, learning_rate 0.2)
from sklearn.ensemble import GradientBoostingClassifier

gbm_clf = GradientBoostingClassifier(n_estimators=1000, learning_rate=0.2, random_state=2024)
gbm_clf.fit(X_train, y_train)
y_pred = gbm_clf.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy:.3f}")        # buku: 0.867
print("Akurasi data latih:", round(gbm_clf.score(X_train, y_train), 3))
show_report(y_test, y_pred)
```

### Hasil dan interpretasi
- Akurasi uji **0,867**, identik dengan pohon tunggal (report sama persis), dan akurasi data latih **1,0**.
- Buku menafsirkan kesamaan ini sebagai tanda keduanya *overfitting* dan "terjebak di maksimum lokal". Tafsiran itu perlu hati-hati: data uji hanya 45 sampel sehingga akurasi identik (39 benar di kedua model) bisa kebetulan. Yang jelas didukung angka adalah akurasi latih 1,0 vs uji 0,867, tanda model terlalu fleksibel untuk data sekecil ini (1000 pohon, `learning_rate=0,2`). Tuning hyperparameter di recipe berikutnya menjawab hal tersebut.

## 4. Hyperparameter Tuning for Trees and Ensembles

### Konsep
- Alat yang sama dengan bab sebelumnya (**grid search + cross-validation**), hanya hyperparameter-nya khas pohon dan ensemble: kedalaman maksimum, jumlah estimator, dan learning rate.
- **Grid search** menguji semua kombinasi; model dengan skor CV terbaik menjadi konfigurasi optimal. **Cross-validation** menilai tiap kombinasi pada beberapa subset data untuk menyimulasikan data baru.
- **`RandomizedSearchCV`** mengambil sampel acak sejumlah `n_iter` kombinasi dari ruang parameter, dan sering lebih cepat untuk ruang besar.

> **Catatan koreksi:** buku menyebut `RandomizedSearchCV` "menyesuaikan hyperparameter di tiap iterasi". Itu tidak benar: tiap kombinasi diambil **independen** secara acak dan tidak belajar dari iterasi sebelumnya. Pencarian adaptif adalah ciri metode lain (misalnya optimasi Bayesian atau *successive halving*).

### Kode

```python
# [Buku] Grid search untuk GradientBoostingClassifier
from sklearn.model_selection import GridSearchCV

param_grid = {
    "n_estimators": [50, 100, 150],
    "max_depth": [3, 4, 5],
    "learning_rate": [0.01, 0.1, 0.2],
}
grid_search = GridSearchCV(GradientBoostingClassifier(random_state=2024),
                           param_grid, cv=5, scoring="accuracy")
grid_search.fit(X_train, y_train)
print(f"Best Parameters: {grid_search.best_params_}")
print("Best CV score  :", round(grid_search.best_score_, 4))
```

```python
# [Buku] Evaluasi model terbaik pada data uji
best_model = grid_search.best_estimator_
y_pred = best_model.predict(X_test)
print(f"Accuracy: {accuracy_score(y_test, y_pred):.3f}")
show_report(y_test, y_pred)
```

```python
# [Buku] Heatmap hasil grid search
import seaborn as sns

results = pd.DataFrame(grid_search.cv_results_)
pivot_table = results.pivot_table(values="mean_test_score",
                                  index="param_max_depth", columns="param_n_estimators")
plt.figure(figsize=(8, 6))
sns.heatmap(pivot_table, annot=True, fmt=".3f", cmap="viridis")
plt.title("Grid Search Accuracy by Max Depth and Number of Estimators")
plt.xlabel("Number of Estimators"); plt.ylabel("Max Depth")
plt.show()
```

### Hasil dan interpretasi
- Parameter terbaik `{'learning_rate': 0.01, 'max_depth': 4, 'n_estimators': 50}` dengan skor CV **0,981**, sama dengan buku. Akurasi uji **0,867** (report identik dengan model sebelumnya; Figure 8.6).
- **20 dari 27 kombinasi seri** di peringkat 1 dengan skor CV yang sama (0,981). `GridSearchCV` hanya memilih yang pertama ditemui, jadi "terbaik" di sini hampir sewenang-wenang.
- **Catatan heatmap:** `pivot_table` secara default **merata-ratakan** `learning_rate` untuk tiap sel (max_depth × n_estimators), jadi heatmap menunjukkan rata-rata tiga learning rate. Kesimpulan buku bahwa akurasi tidak naik di atas `max_depth=4` dan `n_estimators=50` sesuai: seluruh sel kedalaman 4 dan 5 bernilai 0,981, sedangkan kedalaman 3 lebih rendah (0,975 / 0,971 / 0,971).

## 5. Comparing Ensemble Methods

### Konsep
| Metode | Cara kerja | Efek utama |
|---|---|---|
| **Bagging** (random forest) | Model independen pada sampel bootstrap, digabung dengan voting/rata-rata | Menurunkan **variansi** |
| **Boosting** (gradient boosting) | Model berurutan, tiap model memperbaiki kesalahan sebelumnya | Menurunkan **bias** |
| **Stacking** | Beberapa model dasar (*base learners*) beragam; prediksinya menjadi input bagi **meta-model** (`final_estimator`) | Memanfaatkan kekuatan algoritma yang berbeda |

- Pada `StackingClassifier`, meta-model dilatih dari prediksi **out-of-fold** (lewat `cv`) model dasar, agar meta-model tidak belajar dari prediksi yang "melihat" labelnya sendiri.
- Strategi terbaik bergantung pada masalah dan data; bandingkan pada beberapa dataset dan beberapa metrik.

### Kode

```python
# [Buku] Bagging vs boosting vs stacking
from sklearn.ensemble import StackingClassifier
from sklearn.linear_model import LogisticRegression

rf_clf = RandomForestClassifier(n_estimators=100, random_state=2024)
gb_clf = GradientBoostingClassifier(n_estimators=100, random_state=2024)
stacking_clf = StackingClassifier(
    estimators=[("rf", rf_clf), ("gb", gb_clf)],
    final_estimator=LogisticRegression(),
    cv=5)

rf_clf.fit(X_train, y_train)
gb_clf.fit(X_train, y_train)
stacking_clf.fit(X_train, y_train)

rf_accuracy = accuracy_score(y_test, rf_clf.predict(X_test))
gb_accuracy = accuracy_score(y_test, gb_clf.predict(X_test))
stacking_accuracy = accuracy_score(y_test, stacking_clf.predict(X_test))

print(f"Random Forest Accuracy: {rf_accuracy:.2f}")
print(f"Gradient Boosting Accuracy: {gb_accuracy:.2f}")
print(f"Stacking Accuracy: {stacking_accuracy:.2f}")
```

```python
# [Buku] Bar plot perbandingan
methods = ["Random Forest", "Gradient Boosting", "Stacking"]
accuracies = [rf_accuracy, gb_accuracy, stacking_accuracy]

plt.figure(figsize=(8, 6))
plt.bar(methods, accuracies, color=["skyblue", "salmon", "lightgreen"])
plt.title("Accuracy Comparison of Ensemble Methods")
plt.ylabel("Accuracy"); plt.ylim(0, 1)
plt.show()
```

### Hasil dan interpretasi
- Random forest **0,91**, gradient boosting **0,87**, stacking **0,87** (sama dengan buku), yaitu 41, 39, dan 39 benar dari 45 sampel uji.
- Buku menyimpulkan random forest mengungguli yang lain. Selisihnya hanya **2 sampel uji**, sehingga pada satu split sekecil ini belum bermakna statistik. Bandingkan dengan cross-validation berulang dan beberapa dataset sebelum memilih metode (sejalan dengan saran di bagian "There's more" buku).
- Stacking tidak otomatis lebih baik: meta-model hanya menggabungkan prediksi model dasar, dan kemungkinan di sini kedua model dasar sudah sangat mirip pada data Iris yang sederhana sehingga sedikit yang bisa dipelajari dari kombinasinya.
