## 0. Introduction to Distance Metrics
**Metrik jarak** mengukur seberapa mirip (atau berbeda) dua titik data di ruang fitur. Ini dasar bagi klasifikasi, regresi, dan klasterisasi. Tiga metrik utama:

| Metrik | Penjelasan | Rumus (dua titik $x, y$ berdimensi $n$) |
|---|---|---|
| **Euclidean** | Jarak garis lurus ("sejauh burung terbang"). Default di kebanyakan kasus | $d = \sqrt{\sum_{i=1}^{n}(x_i - y_i)^2}$ |
| **Manhattan** (L1 / *city block*) | Jarak menyusuri grid horizontal dan vertikal seperti jalan kota. Lebih tahan terhadap outlier karena memakai selisih absolut, bukan kuadrat | $d = \sum_{i=1}^{n}\lvert x_i - y_i\rvert$ |
| **Minkowski** | Generalisasi keduanya dengan parameter $p$ | $d = \left(\sum_{i=1}^{n}\lvert x_i - y_i\rvert^{p}\right)^{1/p}$ |

Untuk $p=1$ Minkowski menjadi Manhattan, untuk $p=2$ menjadi Euclidean.

```python
# [Tambahan] Hitung tiga metrik jarak secara manual dan bandingkan dengan scikit-learn
from sklearn.metrics import pairwise_distances

a = np.array([[1.0, 2.0]]); b = np.array([[4.0, 6.0]])
print("Euclidean :", np.sqrt(((a - b) ** 2).sum()), "| sklearn:", pairwise_distances(a, b, metric="euclidean")[0, 0])
print("Manhattan :", np.abs(a - b).sum(),            "| sklearn:", pairwise_distances(a, b, metric="manhattan")[0, 0])
print("Minkowski p=3:", round((np.abs(a - b) ** 3).sum() ** (1 / 3), 4),
      "| sklearn:", round(pairwise_distances(a, b, metric="minkowski", p=3)[0, 0], 4))
```

**Interpretasi:** selisih koordinat kedua titik adalah (3, 4). Euclidean = $\sqrt{9+16} = 5$, Manhattan = $3+4 = 7$. Jarak Manhattan selalu $\ge$ Euclidean. Hasil manual sama dengan hasil scikit-learn.

## 1. Understanding KNNs
**KNN** adalah algoritma *supervised* dasar untuk **klasifikasi dan regresi**. Idenya: titik baru diperkirakan karakteristiknya dari titik-titik terdekat yang sudah diketahui di ruang fitur, berdasarkan metrik jarak.

### Getting ready: dataset Iris
150 sampel bunga iris, 4 fitur (panjang/lebar sepal dan petal, dalam cm), 3 kelas. Data dibagi 80% latih dan 20% uji.

```python
# [Dari buku] Muat dataset Iris dan split
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split

iris = load_iris()
X = iris.data
y = iris.target

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=2024)
print("Train:", X_train.shape, "| Test:", X_test.shape)
```

### How to do it…

```python
# [Dari buku] KNN dasar dengan k = 3
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(X_train, y_train)

y_pred = knn.predict(X_test)
print("Accuracy:", accuracy_score(y_test, y_pred))   # buku: 0.9333333333333333
```

**Interpretasi:** dengan pengaturan hampir default, KNN sudah mencapai akurasi **93,3%** pada data uji (28 dari 30 sampel benar), sesuai buku.

### How it works…
KNN mengklasifikasikan titik baru berdasarkan label mayoritas dari $k$ tetangga terdekatnya di data latih:
1. **Memilih $k$:** $k$ kecil membuat model sensitif terhadap *noise* (variansi tinggi, risiko overfitting). $k$ besar menghaluskan prediksi tetapi bisa mengabaikan pola lokal (bias tinggi, risiko underfitting).
2. **Menghitung jarak:** memakai metrik seperti Euclidean, Manhattan, atau Minkowski. Pilihan metrik dapat mengubah hasil klasifikasi.
3. **Voting:** untuk klasifikasi, kelas dengan suara terbanyak di antara $k$ tetangga menjadi prediksi. Untuk **regresi**, prediksi adalah **rata-rata** nilai tetangga.

KNN adalah algoritma **lazy learner**: `fit()` hanya menyimpan data latih, perhitungan jarak baru dilakukan saat `predict()`.

### There's more…: memilih $k$
- **Cross-validation:** evaluasi beberapa nilai $k$ lalu pilih yang errornya terkecil.
- **Elbow method:** plot akurasi terhadap $k$ dan cari titik "siku" di mana kenaikan performa mulai tidak berarti.
- **Variasi metrik jarak:** misalnya *Hamming* untuk data kategorikal.

Buku menjelaskan elbow method tanpa kode, jadi sel berikut **[Tambahan]**.

```python
# [Tambahan] Elbow method: akurasi CV vs k (pada data latih saja, test set tidak disentuh)
from sklearn.model_selection import cross_val_score

ks = range(1, 31)
cv_means = [cross_val_score(KNeighborsClassifier(n_neighbors=k), X_train, y_train, cv=5).mean() for k in ks]

plt.figure(figsize=(8, 4))
plt.plot(list(ks), cv_means, "o-")
plt.xlabel("k (n_neighbors)"); plt.ylabel("Akurasi CV (5-fold)")
plt.title("Elbow method untuk memilih k (Iris)"); plt.grid(True, alpha=0.3); plt.show()
best_k = list(ks)[int(np.argmax(cv_means))]
print("k terbaik menurut CV:", best_k, "| akurasi CV:", round(max(cv_means), 4))
```

**Interpretasi:** akurasi bervariasi mengikuti $k$; pada Iris, banyak nilai $k$ menghasilkan akurasi yang mirip. Karena data kecil, perbedaan antar-$k$ yang kecil sebaiknya tidak ditafsirkan berlebihan. Pilih $k$ yang stabil, bukan sekadar yang tertinggi di satu titik.

**Pemilihan metrik jarak.** Buku menyertakan diagram keputusan (Figure 4.1, berbentuk gambar) untuk memilih metrik berdasarkan karakteristik data. Ringkasnya, dari teks buku: Euclidean sebagai default (fitur kontinu berskala serupa), Manhattan bila selisih absolut lebih bermakna atau data lebih rentan outlier, dan metrik lain seperti Hamming untuk data kategorikal.

## 2. Distance Metrics Overview
Pemilihan metrik jarak memengaruhi bagaimana titik diklasifikasikan dan klaster dibentuk. Recipe ini membandingkan Euclidean, Manhattan, dan Minkowski pada dua dataset sintetis dengan sifat berbeda.

### Getting ready
- **Noisy circles:** dua lingkaran konsentris dengan *noise* tinggi (buku menduga Euclidean cocok).
- **Checkerboard:** titik-titik pada grid dengan pola papan catur (buku menduga Manhattan cocok karena struktur grid).

> **Perbaikan reproduksi:** `make_circles` di buku tidak diberi `random_state`, sehingga hasilnya berubah tiap dijalankan. Saya tambahkan `random_state=2024`. Angka akurasi Anda akan berbeda dari Figure 4.2 buku, dan itu wajar.

```python
# [Dari buku] Dua dataset sintetis (+ random_state agar reproducible)
from sklearn.datasets import make_circles

n_samples = 300

# Circular dataset dengan noise tinggi
X_circles, y_circles = make_circles(n_samples=n_samples, noise=0.3, factor=0.3, random_state=2024)
X_circles_train, X_circles_test, y_circles_train, y_circles_test = train_test_split(
    X_circles, y_circles, test_size=0.2, random_state=2024)

# Checkerboard dataset (grid 17 x 17 = 289 titik)
x = np.linspace(0, 4, int(np.sqrt(n_samples)))
y = np.linspace(0, 4, int(np.sqrt(n_samples)))
xx, yy = np.meshgrid(x, y)
X_moons = np.column_stack((xx.ravel(), yy.ravel()))
y_moons = np.mod(np.floor(xx.ravel()) + np.floor(yy.ravel()), 2)
X_moons_train, X_moons_test, y_moons_train, y_moons_test = train_test_split(
    X_moons, y_moons, test_size=0.2, random_state=2024)

print("Circles:", X_circles.shape, "| Checkerboard:", X_moons.shape)
```

### How to do it…
Untuk setiap metrik, bangun dua classifier KNN ($k=3$) lalu ukur akurasi pada masing-masing dataset.

```python
# [Dari buku] Bandingkan metrik jarak pada kedua dataset
metrics = ["euclidean", "manhattan", "minkowski"]
results = {"circles": {}, "moons": {}}

for metric in metrics:
    knn_circles = KNeighborsClassifier(n_neighbors=3, metric=metric)
    knn_circles.fit(X_circles_train, y_circles_train)
    y_pred_circles = knn_circles.predict(X_circles_test)
    results["circles"][metric] = accuracy_score(y_circles_test, y_pred_circles)

    knn_moons = KNeighborsClassifier(n_neighbors=3, metric=metric)
    knn_moons.fit(X_moons_train, y_moons_train)
    y_pred_moons = knn_moons.predict(X_moons_test)
    results["moons"][metric] = accuracy_score(y_moons_test, y_pred_moons)

display(pd.DataFrame(results).round(4))
```

```python
# [Dari buku] Visualisasi dataset dan akurasi per metrik
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 5))

scatter1_train = ax1.scatter(X_circles_train[:, 0], X_circles_train[:, 1], c=y_circles_train,
                             cmap="viridis", alpha=0.6, label="Training Points")
scatter1_test = ax1.scatter(X_circles_test[:, 0], X_circles_test[:, 1], color="red", marker="x", s=50,
                            label="Test Points")
ax1.set_title("Noisy Circles Dataset")
lines1 = [plt.Line2D([0], [0], color="white") for _ in metrics]
legend_text1 = [
    f"Euclidean: {results['circles']['euclidean']:.3f}",
    f"Manhattan: {results['circles']['manhattan']:.3f}",
    f"Minkowski: {results['circles']['minkowski']:.3f}",
]
ax1.legend(lines1 + [scatter1_train, scatter1_test], legend_text1 + ["Training Points", "Test Points"],
           title="Accuracy by Metric", bbox_to_anchor=(1.05, 1), loc="upper left")

scatter2_train = ax2.scatter(X_moons_train[:, 0], X_moons_train[:, 1], c=y_moons_train,
                             cmap="viridis", alpha=0.6, label="Training Points")
scatter2_test = ax2.scatter(X_moons_test[:, 0], X_moons_test[:, 1], color="red", marker="x", s=50,
                            label="Test Points")
ax2.set_title("Checkerboard Dataset")
lines2 = [plt.Line2D([0], [0], color="white") for _ in metrics]
legend_text2 = [
    f"Euclidean: {results['moons']['euclidean']:.3f}",
    f"Manhattan: {results['moons']['manhattan']:.3f}",
    f"Minkowski: {results['moons']['minkowski']:.3f}",
]
ax2.legend(lines2 + [scatter2_train, scatter2_test], legend_text2 + ["Training Points", "Test Points"],
           title="Accuracy by Metric", bbox_to_anchor=(1.05, 1), loc="upper left")

plt.tight_layout()
plt.show()
```

```python
# [Tambahan] Bukti bahwa 'euclidean' dan 'minkowski' (p=2 default) identik, dan Minkowski p=1 = Manhattan
for name, (Xtr, ytr, Xte, yte) in {
    "circles": (X_circles_train, y_circles_train, X_circles_test, y_circles_test),
    "checkerboard": (X_moons_train, y_moons_train, X_moons_test, y_moons_test),
}.items():
    pred = {}
    for label, kw in {"euclidean": dict(metric="euclidean"),
                      "minkowski(p=2, default)": dict(metric="minkowski"),
                      "manhattan": dict(metric="manhattan"),
                      "minkowski(p=1)": dict(metric="minkowski", p=1)}.items():
        pred[label] = KNeighborsClassifier(n_neighbors=3, **kw).fit(Xtr, ytr).predict(Xte)
    print(name,
          "| euclidean == minkowski(p=2):", np.array_equal(pred["euclidean"], pred["minkowski(p=2, default)"]),
          "| manhattan == minkowski(p=1):", np.array_equal(pred["manhattan"], pred["minkowski(p=1)"]))
```

**Interpretasi:** pada setiap dataset, akurasi `euclidean` dan `minkowski` **selalu sama** karena Minkowski dengan $p=2$ (default scikit-learn) *adalah* Euclidean, sedangkan `manhattan` setara Minkowski dengan $p=1$ (sel pembuktian di atas). Jadi dari tiga metrik pada buku, sebenarnya hanya ada dua yang berbeda secara nyata. Pada hasil saya, di *noisy circles* Euclidean unggul tipis (0,783 vs 0,767; selisih hanya 1 dari 60 sampel uji), sedangkan di buku justru Manhattan yang sedikit unggul. Perbedaan arah ini menunjukkan bahwa selisihnya hanya efek acak data/split, bukan bukti satu metrik lebih baik (buku sendiri menyebutnya "mungkin hanya kebetulan"). Pada *checkerboard*, Manhattan unggul (0,845 vs 0,810; selisih 2 dari 58 sampel uji). Ini sejalan dengan dugaan buku bahwa struktur grid selaras dengan jarak berbasis sumbu, tetapi selisih sekecil itu belum cukup kuat untuk dijadikan bukti.

### How it works…
- **Euclidean:** jarak garis lurus; paling umum untuk KNN. Sensitif terhadap outlier dan paling baik bila fitur berskala serupa.
- **Manhattan:** jumlah selisih absolut koordinat; kurang sensitif terhadap outlier dibanding Euclidean.
- **Minkowski:** generalisasi keduanya; parameter `p` mengganti metrik ($p=1$ Manhattan, $p=2$ Euclidean).

**Dampak pada performa model:**
- **Sensitif terhadap skala:** Euclidean sangat dipengaruhi skala fitur, sehingga standardisasi/normalisasi sebelum KNN penting.
- **Efek dimensi:** pada ruang berdimensi tinggi, jarak menjadi kurang bermakna (*curse of dimensionality*); beberapa metrik bisa lebih tahan daripada yang lain.

**Saran praktis:** pertimbangkan sifat data (kontinu vs kategorikal), coba beberapa metrik, dan bandingkan dengan cross-validation.

```python
# [Tambahan] Pentingnya scaling pada KNN: dataset Wine (fitur berskala sangat berbeda)
from sklearn.datasets import load_wine
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

wine = load_wine()
raw = KNeighborsClassifier(n_neighbors=5)
scaled = Pipeline([("sc", StandardScaler()), ("knn", KNeighborsClassifier(n_neighbors=5))])
print("Wine - KNN tanpa scaling   :", round(cross_val_score(raw, wine.data, wine.target, cv=5).mean(), 4))
print("Wine - KNN dengan scaling  :", round(cross_val_score(scaled, wine.data, wine.target, cv=5).mean(), 4))
```

**Interpretasi:** pada Wine, fitur seperti `proline` berskala ratusan-ribuan sehingga mendominasi jarak Euclidean. Setelah standardisasi, semua fitur berkontribusi adil dan akurasi KNN naik tajam. (Pada Iris, semua fitur sudah dalam cm dan berskala serupa sehingga masalah ini tidak terlihat.)

## 3. Hyperparameter Tuning in KNN
**Parameter** dipelajari model dari data saat pelatihan (misalnya kemiringan dan intersep pada regresi linear). **Hyperparameter** ditetapkan sebelum pelatihan dan mengatur *bagaimana* model belajar (misalnya $k$, `weights`, `metric` pada KNN).

### Getting ready
Dataset Iris yang sama. Yang akan dituning: jumlah tetangga (`n_neighbors`), apakah tetangga diberi bobot berdasarkan jarak (`weights`), dan metrik jarak (`metric`).

### How to do it…
**Grid search** mencoba **semua kombinasi** hyperparameter di grid, masing-masing dievaluasi dengan **cross-validation**, lalu mengembalikan kombinasi terbaik.

```python
# [Dari buku] GridSearchCV untuk KNN
from sklearn.model_selection import GridSearchCV

knn = KNeighborsClassifier()
param_grid = {
    "n_neighbors": [3, 5, 7, 9, 11],
    "weights": ["uniform", "distance"],
    "metric": ["euclidean", "manhattan"],
}
grid_search = GridSearchCV(knn, param_grid, cv=5, scoring="accuracy")
grid_search.fit(X_train, y_train)

print("Best parameters:", grid_search.best_params_)
print("Best cross-validation score:", grid_search.best_score_)
# buku: {'metric': 'euclidean', 'n_neighbors': 9, 'weights': 'uniform'} dan 0.9916666666666668
```

**Interpretasi:** `GridSearchCV` menguji 5 × 2 × 2 = **20 kombinasi**, masing-masing dengan 5-fold CV (total 100 pelatihan), lalu memilih kombinasi dengan skor CV rata-rata tertinggi (`best_params_`). Dibandingkan KNN default ($k=3$), tuning dapat memberi hasil yang sedikit lebih baik atau sama.

```python
# [Tambahan] Lihat 5 kombinasi teratas dan jumlah kombinasi yang seri (banyak kombinasi bisa menghasilkan skor yang sama)
res = pd.DataFrame(grid_search.cv_results_)[["params", "mean_test_score", "std_test_score", "rank_test_score"]]
display(res.sort_values("rank_test_score").head(8).reset_index(drop=True))
print("Jumlah kombinasi dengan peringkat 1:", int((res["rank_test_score"] == 1).sum()))
```

**Interpretasi:** pada dataset sekecil Iris, beberapa kombinasi sering menghasilkan skor yang sama atau nyaris sama (pada hasil ini ada 2 kombinasi yang seri di peringkat 1 dengan skor 0,9917), sehingga "yang terbaik" ditentukan oleh perbedaan sangat kecil (dan oleh aturan pemecah seri scikit-learn, yaitu kombinasi pertama yang ditemui). Jangan terlalu yakin bahwa kombinasi terpilih jauh lebih baik daripada kombinasi lain di dekatnya.

### How it works…
- **Grid search:** menelusuri seluruh grid secara menyeluruh untuk menemukan kombinasi terbaik berdasarkan performa.
- **Cross-validation:** model dievaluasi pada beberapa pembagian data berbeda sehingga estimasi performa lebih andal dan risiko overfitting pada satu split berkurang.

**Setelah grid search:** (1) latih model akhir dengan parameter terbaik lalu evaluasi pada **test set** terpisah, (2) pilih metrik sesuai tugas (akurasi untuk klasifikasi, MSE untuk regresi).

### There's more…: `RandomizedSearchCV`
Untuk ruang parameter besar, `RandomizedSearchCV` mengambil sampel acak kombinasi (bukan semua), menghemat waktu dengan hasil yang sering sama baiknya.

```python
# [Tambahan] RandomizedSearchCV: hanya 8 dari ruang pencarian yang lebih besar
from sklearn.model_selection import RandomizedSearchCV

param_dist = {
    "n_neighbors": list(range(1, 31)),
    "weights": ["uniform", "distance"],
    "metric": ["euclidean", "manhattan", "chebyshev"],
}
rand_search = RandomizedSearchCV(KNeighborsClassifier(), param_dist, n_iter=8, cv=5, random_state=2024)
rand_search.fit(X_train, y_train)
print("Best parameters:", rand_search.best_params_)
print("Best CV score  :", round(rand_search.best_score_, 4))
print("Ruang pencarian penuh =", 30 * 2 * 3, "kombinasi; yang dicoba =", 8)
```

## 4. Evaluating KNN Performance
Evaluasi menunjukkan seberapa baik model memprediksi dan di mana perlu diperbaiki. Recipe ini memakai tiga pendekatan: **cross-validation**, **learning curve**, dan **confusion matrix + classification report**, dengan model terbaik dari grid search.

> **Perbaikan reproduksi:** buku hanya mengimpor `learning_curve`, `confusion_matrix`, dan `seaborn` di recipe ini, tetapi kodenya juga memakai `cross_val_score` dan `classification_report`. Keduanya saya impor secara eksplisit.

```python
# [Dari buku] Impor untuk evaluasi (+ cross_val_score dan classification_report yang terlewat di buku)
from sklearn.model_selection import learning_curve, cross_val_score
from sklearn.metrics import confusion_matrix, classification_report
import seaborn as sns
```

### 4a. Cross-validation scores
Skor akurasi dari 5 fold pada data latih memakai model terbaik hasil grid search.

```python
# [Dari buku] Skor cross-validation
cv_scores = cross_val_score(grid_search.best_estimator_, X_train, y_train, cv=5)
print("Cross-validation scores:", cv_scores)
print("Mean CV score:", cv_scores.mean())
print("Standard deviation:", cv_scores.std())
```

**Interpretasi:** setiap angka adalah akurasi pada satu fold. Rata-rata menggambarkan performa umum, dan standar deviasi menggambarkan kestabilan antar-fold (semakin kecil, semakin stabil).

### 4b. Learning curve
Kurva yang memplot skor model (pelatihan dan validasi) terhadap ukuran data latih. Berguna untuk mendiagnosis *underfitting*/*overfitting* dan melihat apakah menambah data masih bermanfaat.

```python
# [Dari buku] Learning curve
train_sizes, train_scores, test_scores = learning_curve(
    grid_search.best_estimator_, X_train, y_train,
    train_sizes=np.linspace(0.1, 1.0, 10), cv=5)

plt.figure(figsize=(10, 6))
plt.plot(train_sizes, np.mean(train_scores, axis=1), label="Training score")
plt.plot(train_sizes, np.mean(test_scores, axis=1), label="Cross-validation score")
plt.xlabel("Training examples")
plt.ylabel("Score")
plt.legend(loc="best")
plt.title("Learning Curves")
plt.show()
```

**Cara membaca:** bila skor latih dan validasi sama-sama tinggi dan saling mendekat, model cocok dan menambah data tidak banyak membantu. Bila skor latih tinggi tetapi validasi jauh lebih rendah, ada indikasi overfitting. Bila keduanya rendah, indikasi underfitting. Kurva yang sudah mendatar menandakan tambahan data tidak lagi banyak meningkatkan performa (menghemat biaya komputasi).

**Pada hasil ini:** skor validasi naik cepat sampai sekitar 20–50 sampel, lalu mendekati skor latih (±0,99) pada ukuran data penuh. Jarak antara kedua kurva hampir hilang, jadi tidak ada tanda overfitting yang berarti, dan menambah data pada Iris kemungkinan tidak banyak membantu.

### 4c. Confusion matrix
Ringkasan visual prediksi: baris = label sebenarnya, kolom = label prediksi. Angka pada diagonal adalah prediksi benar.

```python
# [Dari buku] Confusion matrix
y_pred = grid_search.best_estimator_.predict(X_test)
class_names = iris.target_names

cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
            xticklabels=class_names, yticklabels=class_names,
            cbar_kws={"label": "Number of Samples"})
plt.xlabel("Predicted")
plt.ylabel("True")
plt.title("Confusion Matrix")
plt.show()
print("Akurasi test (model terbaik):", accuracy_score(y_test, y_pred))
```

**Interpretasi:** nilai besar pada diagonal berarti sebagian besar sampel diklasifikasikan benar. Pada hasil ini 28 dari 30 sampel benar: seluruh 12 *setosa* benar, 1 *versicolor* salah diprediksi sebagai *virginica*, dan 1 *virginica* salah diprediksi sebagai *versicolor* (dua kelas yang memang mirip).

### 4d. Classification report
Tabel metrik per kelas.

```python
# [Dari buku] Classification report -> DataFrame berwarna
report_dict = classification_report(y_test, y_pred, target_names=class_names, output_dict=True)
report_df = pd.DataFrame(report_dict).transpose()

styled_df = (report_df.style
    .background_gradient(cmap="Blues", subset=["precision", "recall", "f1-score"])
    .format({"precision": "{:.3f}", "recall": "{:.3f}", "f1-score": "{:.3f}", "support": "{:.0f}"}))

print("\nClassification Report:")
display(styled_df)
```

**Hasil pada eksekusi ini:** *setosa* precision/recall/F1 = 1,000; *versicolor* 0,875; *virginica* 0,900; accuracy 0,933 (30 sampel uji). Makin tinggi nilai makin baik, dan *support* adalah jumlah sampel uji per kelas.

> **Catatan buku:** biasanya kita curiga bila ada precision/recall/F1 = 100%, tetapi kelas *setosa* di Iris memang sangat mudah dipisahkan. Selain itu, "performa bagus" selalu harus dinilai terhadap masalah bisnis yang dimodelkan, bukan sekadar dari angkanya.

### Memahami metrik evaluasi
| Metrik | Arti | Rumus |
|---|---|---|
| **Learning curve** | Skor vs ukuran data latih; deteksi under/overfitting | (grafik) |
| **Confusion matrix** | Hitungan TP, TN, FP, FN per kelas | (tabel) |
| **Precision** | Dari yang diprediksi positif, berapa yang benar | $\dfrac{TP}{TP + FP}$ |
| **Recall** (sensitivity) | Dari semua yang sebenarnya positif, berapa yang tertangkap | $\dfrac{TP}{TP + FN}$ |
| **F1-score** | Rata-rata harmonik precision dan recall; berguna pada data tidak seimbang | $\dfrac{2 \times Precision \times Recall}{Precision + Recall}$ |
| **Accuracy** | Proporsi prediksi benar dari seluruh prediksi | $\dfrac{TP + TN}{TP + TN + FP + FN}$ |
| **Macro avg** | Rata-rata metrik antar kelas (tiap kelas bobot sama) | — |
| **Weighted avg** | Rata-rata metrik antar kelas, diberi bobot jumlah sampel kelas | — |

(TP = true positive, TN = true negative, FP = false positive, FN = false negative.)

### There's more…: data tidak seimbang (*imbalanced*)
Bila satu kelas jauh lebih banyak daripada yang lain, akurasi saja menyesatkan. Gunakan **precision-recall curve** atau **ROC-AUC**, dan pertimbangkan **resampling** (*oversampling* kelas minoritas atau *undersampling* kelas mayoritas) sebelum pelatihan.

```python
# [Tambahan] Mengapa akurasi saja menyesatkan pada data tidak seimbang
from sklearn.datasets import make_classification
from sklearn.metrics import precision_score, recall_score, f1_score

Xi, yi = make_classification(n_samples=1000, weights=[0.95, 0.05], flip_y=0, class_sep=0.8, random_state=2024)
Xi_tr, Xi_te, yi_tr, yi_te = train_test_split(Xi, yi, test_size=0.2, stratify=yi, random_state=2024)
pi = KNeighborsClassifier(n_neighbors=5).fit(Xi_tr, yi_tr).predict(Xi_te)
print("Proporsi kelas minoritas di test:", round(yi_te.mean(), 3))
print("Akurasi  :", round(accuracy_score(yi_te, pi), 3), "<- tampak bagus")
print("Precision:", round(precision_score(yi_te, pi, zero_division=0), 3),
      "| Recall:", round(recall_score(yi_te, pi), 3),
      "| F1:", round(f1_score(yi_te, pi), 3), "<- kelas minoritas sebenarnya sulit")
```

**Interpretasi:** model yang menebak "kelas mayoritas" saja sudah akurat ~95%. Pada simulasi ini, precision dan recall kelas minoritas bahkan **0**: KNN tidak menemukan satu pun kasus minoritas, padahal kelas itu biasanya justru yang penting (misalnya deteksi penipuan). Akurasi 95% tampak bagus tetapi menyembunyikan kegagalan total pada kelas minoritas. Karena itu, pada data tidak seimbang gunakan precision, recall, F1, atau ROC-AUC.

## Kesimpulan Chapter 4
- **Metrik jarak** (Euclidean, Manhattan, Minkowski) mengukur kemiripan; Minkowski dengan $p=2$ sama dengan Euclidean dan $p=1$ sama dengan Manhattan.
- **KNN** mengklasifikasikan lewat voting $k$ tetangga terdekat (regresi: rata-rata). Pemilihan $k$ mengatur trade-off bias-variansi, dan KNN adalah *lazy learner*.
- **Scaling** sangat penting untuk KNN karena jarak dipengaruhi skala fitur.
- **Hyperparameter tuning** dengan `GridSearchCV` (semua kombinasi) atau `RandomizedSearchCV` (sampel acak), selalu dengan cross-validation.
- **Evaluasi** yang lengkap: cross-validation, learning curve, confusion matrix, dan classification report (precision, recall, F1). Pada data tidak seimbang, akurasi saja tidak cukup.
